```python
import os
import geopandas as gpd
import fiona
import matplotlib.pyplot as plt
import mplleaflet
import numpy as np
import time
import csv

kml_file = 'olek1_v1.kml'
gpd.io.file.fiona.drvsupport.supported_drivers['KML'] = 'rw'
df = gpd.read_file(kml_file, driver='KML')
ax = df.plot()

lats = []
longs = []

with open('../raw_log.csv', mode='r') as infile:
    reader = csv.DictReader(infile)

    for row in reader:
        lats.append(float(row['field.latitude']))
        longs.append(float(row['field.longitude']))

# longs = longs[0:10]
# lats = lats[0:10]

ax.plot(longs[0:10], lats[0:10], 'r.')
ax.plot(longs[0:10], lats[0:10], 'b')

mplleaflet.display(fig=ax.figure)
# mplleaflet.show()

# mplleaflet.display(fig=ax.figure, crs=df.crs)

# ax.plot(longs, lats, 'r.')
# ax.plot(longs, lats, 'b')

# plt.draw()
# plt.pause(0.0001)
# ax.cla()

```
