```python
import rosbag

MAG_VAR = 1e-14
ANG_VEL_VAR = 0.001
LIN_ACC_VAR = 0.001

input_bag_path = '/home/olek/projects/bags/mag_tuning/outdoor_test_1_filtered.bag'
output_bag_path = '/home/olek/projects/bags/mag_tuning/edited.bag'

with rosbag.Bag(output_bag_path, 'w') as outbag:
    with rosbag.Bag(input_bag_path) as inbag:
        for topic, msg, t in inbag.read_messages():
            if 'agribot/magnetic_field' in topic:
                msg.magnetic_field_covariance = [MAG_VAR, 0.0, 0.0, 0.0, MAG_VAR, 0.0, 0.0, 0.0, MAG_VAR]
            elif 'agribot/imu/raw' in topic:
                msg.angular_velocity_covariance = [ANG_VEL_VAR, 0.0, 0.0, 0.0, ANG_VEL_VAR, 0.0, 0.0, 0.0, ANG_VEL_VAR]
                msg.linear_acceleration_covariance = [LIN_ACC_VAR, 0.0, 0.0, 0.0, LIN_ACC_VAR, 0.0, 0.0, 0.0, LIN_ACC_VAR]
            outbag.write(topic, msg, t)

        outbag.flush()

        print(inbag.get_message_count())
        print(outbag.get_message_count())
```
